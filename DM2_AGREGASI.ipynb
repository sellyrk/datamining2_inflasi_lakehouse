```python
!pip install -q duckdb pyarrow openpyxl

import re
import duckdb
import numpy as np
import pandas as pd
from pathlib import Path

# ====== KONFIGURASI (sesuaikan) ======
BASE  = Path("/content/drive/MyDrive/DM2_INFLASI")        # mis. Path("/content/drive/MyDrive/bps_2025")
RAW   = BASE / "/content/drive/MyDrive/DM2_INFLASI/data_lake"            # folder berisi semua .xlsx (subfolder boleh)
LAKE  = BASE / "/content/drive/MyDrive/DM2_INFLASI/data_lake/raw"           # output parquet per domain
DB    = BASE / "bps_2025.duckdb"
TAHUN = 2025

LAKE.mkdir(parents=True, exist_ok=True)
con = duckdb.connect(str(DB))

XLSX = {p.name: p for p in RAW.rglob("*.xlsx")}
print(len(XLSX), "file xlsx ditemukan")

def find(keyword):
    hits = [p for n, p in XLSX.items() if keyword.lower() in n.lower()]
    if len(hits) != 1:
        raise FileNotFoundError(f"'{keyword}' cocok dengan {len(hits)} file: {[h.name for h in hits]}")
    return hits[0]

BULAN_NAMA = ["Januari", "Februari", "Maret", "April", "Mei", "Juni", "Juli",
              "Agustus", "September", "Oktober", "November", "Desember"]
BULAN_MAP = {}
for i, n in enumerate(BULAN_NAMA, 1):
    BULAN_MAP[n.lower()] = i
    BULAN_MAP[n.lower()[:3]] = i
BULAN_MAP["ags"] = 8

ALIAS = {
    "SEMUA PROVINSI": "INDONESIA",
    "DAERAH ISTIMEWA YOGYAKARTA": "DI YOGYAKARTA",
    "D I YOGYAKARTA": "DI YOGYAKARTA",
    "DAERAH KHUSUS IBUKOTA JAKARTA": "DKI JAKARTA",
    "BANGKA BELITUNG": "KEPULAUAN BANGKA BELITUNG",
}

def norm_name(s):
    """Samakan penulisan nama provinsi: 'PROV ACEH', '11 Aceh', 'KEP. RIAU' -> kunci seragam."""
    s = str(s).upper()
    s = re.sub(r"^\s*\d+\s+", "", s)
    s = re.sub(r"^\s*(PROVINSI|PROV)\b\.?\s*", "", s)
    s = re.sub(r"\bKEP\b\.?", "KEPULAUAN", s)
    s = re.sub(r"[^A-Z0-9 ]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return ALIAS.get(s, s)

def to_num(s, thousands=","):
    """'12,750' -> 12750.0 ; '-' / '–' -> NaN"""
    if pd.api.types.is_numeric_dtype(s):
        return s.astype(float)
    x = s.astype(str).str.strip().str.replace(thousands, "", regex=False)
    x = x.replace({"-": np.nan, "–": np.nan, "—": np.nan, "": np.nan,
                   "nan": np.nan, "NaN": np.nan, "None": np.nan})
    return pd.to_numeric(x, errors="coerce")

def melt_bulan(df, value_name):
    """Tabel lebar (kolom = nama bulan) -> panjang (nama_wilayah, bulan_id, nilai)."""
    idc = df.columns[0]
    bcols = [c for c in df.columns[1:] if str(c).strip().lower() in BULAN_MAP]
    long = df.melt(id_vars=idc, value_vars=bcols, var_name="bulan_nama", value_name=value_name)
    bulan = long["bulan_nama"].astype(str).str.strip().str.lower().map(BULAN_MAP)
    long["bulan_id"] = TAHUN * 100 + bulan
    long[value_name] = to_num(long[value_name])
    long = long.rename(columns={idc: "nama_wilayah"})
    return long[["nama_wilayah", "bulan_id", value_name]].dropna(subset=[value_name])

def cek_unik(df, keys, nama):
    print(f"[cek] {nama}: {df.duplicated(subset=keys).sum()} duplikat pada kunci {keys}")

def save_table(df, domain, name):
    out = LAKE / domain
    out.mkdir(parents=True, exist_ok=True)
    df.to_parquet(out / f"{name}.parquet", index=False)
    con.register("_tmp", df)
    con.execute(f"CREATE OR REPLACE TABLE {name} AS SELECT * FROM _tmp")
    con.unregister("_tmp")
    print(f"{name:28s} {len(df):>7,} baris  ->  lake/{domain}/{name}.parquet")
```

```python
ump = pd.read_excel(find("Upah Minimum Provinsi"))
kn = ump.iloc[:, 0].astype(str).str.extract(r"^\s*(\d+)\s+(.+?)\s*$").dropna()
kn.columns = ["kode", "nama"]

WILAYAH = {1: "Sumatera", 2: "Sumatera", 3: "Jawa", 5: "Bali & Nusa Tenggara",
           6: "Kalimantan", 7: "Sulawesi", 8: "Maluku", 9: "Papua"}

dim_provinsi = pd.DataFrame({
    "kode_provinsi": kn["kode"].astype(int).values,
    "nama_provinsi": kn["nama"].str.strip().values,
}).sort_values("kode_provinsi")
dim_provinsi = pd.concat(
    [pd.DataFrame({"kode_provinsi": [0], "nama_provinsi": ["INDONESIA"]}), dim_provinsi],
    ignore_index=True)
dim_provinsi["wilayah"] = dim_provinsi["kode_provinsi"].map(
    lambda k: "Nasional" if k == 0 else WILAYAH.get(k // 10, "Lainnya"))
dim_provinsi["nama_key"] = dim_provinsi["nama_provinsi"].map(norm_name)

print("Jumlah provinsi (tanpa nasional):", len(dim_provinsi) - 1, "(harapan: 38)")
KEY2KODE = dict(zip(dim_provinsi["nama_key"], dim_provinsi["kode_provinsi"]))

def add_kode(df, col="nama_wilayah", sumber="", verbose=True):
    """Ganti nama wilayah dengan kode_provinsi; baris yang tidak cocok (catatan kaki, kab/kota) dibuang."""
    kode = df[col].map(norm_name).map(KEY2KODE)
    if verbose:
        tak = sorted(df.loc[kode.isna() & df[col].notna(), col].astype(str).unique())
        if tak:
            print(f"[{sumber}] {len(tak)} nama tidak cocok, dibuang: {tak[:6]}{' ...' if len(tak) > 6 else ''}")
    out = df.assign(kode_provinsi=kode)
    out = out[out["kode_provinsi"].notna()].astype({"kode_provinsi": "int"})
    return out.drop(columns=[col])

dim_waktu = pd.DataFrame({
    "bulan_id": [TAHUN * 100 + m for m in range(1, 13)],
    "tahun": TAHUN,
    "bulan": range(1, 13),
    "nama_bulan": BULAN_NAMA,
    "kuartal": [(m - 1) // 3 + 1 for m in range(1, 13)],
    "semester": [(m - 1) // 6 + 1 for m in range(1, 13)],
})

save_table(dim_provinsi.drop(columns="nama_key"), "Dimensi", "dim_provinsi")
save_table(dim_waktu, "Dimensi", "dim_waktu")
```

```python
frames = []
for p in sorted(RAW.rglob("*Komoditas_*.xlsx")):
    komoditas = p.stem.split("Komoditas_")[-1].strip()
    df = pd.read_excel(p)
    df = df.rename(columns={df.columns[1]: "nama_wilayah"})
    long = df.melt(id_vars="nama_wilayah", value_vars=list(df.columns[2:]),
                   var_name="tanggal", value_name="harga_rp")
    txt = long["tanggal"].astype(str).str.replace(" ", "", regex=False)   # '01/ 01/ 2025' -> '01/01/2025'
    tgl = pd.to_datetime(txt, format="%d/%m/%Y", errors="coerce")
    long["tanggal_data"] = tgl.fillna(pd.to_datetime(long["tanggal"], errors="coerce"))
    long["harga_rp"] = to_num(long["harga_rp"])
    long["nama_komoditas"] = komoditas
    frames.append(long.drop(columns="tanggal"))

harga = pd.concat(frames, ignore_index=True).dropna(subset=["harga_rp", "tanggal_data"])
harga["bulan_id"] = harga["tanggal_data"].dt.year * 100 + harga["tanggal_data"].dt.month

dim_komoditas = pd.DataFrame({"nama_komoditas": sorted(harga["nama_komoditas"].unique())})
dim_komoditas.insert(0, "komoditas_id", range(1, len(dim_komoditas) + 1))
harga = harga.merge(dim_komoditas, on="nama_komoditas")

fact_harga = add_kode(harga, sumber="harga")[
    ["kode_provinsi", "komoditas_id", "bulan_id", "tanggal_data", "harga_rp"]]
cek_unik(fact_harga, ["kode_provinsi", "komoditas_id", "bulan_id"], "fact_harga")

save_table(dim_komoditas, "Pangan", "dim_komoditas")
save_table(fact_harga, "Pangan", "fact_harga")
```

```python
tpt = pd.read_excel(find("Tingkat Pengangguran"))
fact_tpt = add_kode(melt_bulan(tpt, "tpt_pct"), sumber="tpt")
cek_unik(fact_tpt, ["kode_provinsi", "bulan_id"], "fact_tpt")

tpak = pd.read_excel(find("Angkatan Kerja"))
fact_tpak = add_kode(melt_bulan(tpak, "tpak_pct"), sumber="tpak")
cek_unik(fact_tpak, ["kode_provinsi", "bulan_id"], "fact_tpak")

fact_ump = pd.DataFrame({
    "kode_provinsi": kn["kode"].astype(int).values,
    "tahun": TAHUN,
    "ump_rp": to_num(ump.iloc[:, 1].loc[kn.index]).values,
})

# Rata-rata upah: per SEKTOR (nasional), bukan per provinsi
up = pd.read_excel(find("Rata-Rata Upah"))
fact_upah = melt_bulan(up, "upah_rp").rename(columns={"nama_wilayah": "sektor"})

save_table(fact_tpt, "Ekonomi", "fact_tpt")
save_table(fact_tpak, "Ekonomi", "fact_tpak")
save_table(fact_ump, "Ekonomi", "fact_ump")
save_table(fact_upah, "Ekonomi", "fact_upah_sektor_nasional")
```

```python
raw_e = pd.read_excel(find("Inflasi Bulanan"), header=None)

print("Cuplikan struktur file (12 baris x 8 kolom pertama):")
display(raw_e.iloc[:12, :8])

# Baris header bulan = baris dengan nama bulan terbanyak
n_bulan = raw_e.apply(
    lambda r: r.astype(str).str.strip().str.lower().map(BULAN_MAP).notna().sum(), axis=1)
mrow = int(n_bulan.idxmax())
SUB_OFFSET = 1        # label subkelompok ada di baris (mrow - SUB_OFFSET); ubah jika berbeda

sub_label = raw_e.iloc[mrow - SUB_OFFSET].ffill()
bulan_row = raw_e.iloc[mrow].astype(str).str.strip().str.lower().map(BULAN_MAP)

body = raw_e.iloc[mrow + 1:]
prov_col = int(body.apply(lambda c: c.astype(str).str.contains("ACEH", case=False).any()).idxmax())

rows = []
for j in raw_e.columns:
    if j > prov_col and pd.notna(bulan_row[j]):
        rows.append(pd.DataFrame({
            "nama_wilayah": body[prov_col].values,
            "subkelompok": str(sub_label[j]).strip(),
            "bulan_id": TAHUN * 100 + int(bulan_row[j]),
            "inflasi_mtm_pct": to_num(body[j]).values,
        }))

fact_inflasi_energi = add_kode(
    pd.concat(rows, ignore_index=True).dropna(subset=["inflasi_mtm_pct"]),
    sumber="inflasi_energi")
cek_unik(fact_inflasi_energi, ["kode_provinsi", "subkelompok", "bulan_id"], "fact_inflasi_energi")
print(fact_inflasi_energi["subkelompok"].value_counts())

save_table(fact_inflasi_energi, "Energi", "fact_inflasi_energi")
```

```python
# Penduduk miskin (Maret & September)
ms = pd.read_excel(find("Penduduk Miskin"))
ms = ms.rename(columns={ms.columns[0]: "nama_wilayah"})
frames = []
for nama, m in [("Maret", 3), ("September", 9)]:
    frames.append(pd.DataFrame({
        "nama_wilayah": ms["nama_wilayah"],
        "bulan_id": TAHUN * 100 + m,
        "garis_kemiskinan_rp": to_num(ms[f"Garis Kemiskinan - {nama} (Rp)"]),
        "jumlah_miskin_ribu": to_num(ms[f"Jumlah Penduduk Miskin - {nama} (ribu) (Ribu)"]),
        "persen_miskin": to_num(ms[f"Persentase Penduduk Miskin - {nama}"]),
    }))
fact_miskin = add_kode(pd.concat(frames, ignore_index=True), sumber="miskin")
fact_miskin = fact_miskin.dropna(subset=["persen_miskin", "jumlah_miskin_ribu"], how="all")
cek_unik(fact_miskin, ["kode_provinsi", "bulan_id"], "fact_penduduk_miskin")

# Penduduk (tahunan)
pdk = pd.read_excel(find("Laju Pertumbuhan")).iloc[:, :6]
pdk.columns = ["nama_wilayah", "jumlah_penduduk_ribu", "laju_pertumbuhan_pct",
               "persen_penduduk", "kepadatan_km2", "rasio_jenis_kelamin"]
for c in pdk.columns[1:]:
    pdk[c] = to_num(pdk[c])
pdk["tahun"] = TAHUN
fact_penduduk = add_kode(pdk, sumber="penduduk").dropna(subset=["jumlah_penduduk_ribu"])
cek_unik(fact_penduduk, ["kode_provinsi", "tahun"], "fact_penduduk")

# Pengeluaran per kapita (file berisi provinsi + ratusan kab/kota; hanya baris provinsi dipakai)
pk = pd.read_excel(find("Pengeluaran per Kapita"), header=None).iloc[:, :3]
pk.columns = ["nama_wilayah", "pengeluaran_laki_rp", "pengeluaran_perempuan_rp"]
for c in pk.columns[1:]:
    pk[c] = to_num(pk[c])
pk["tahun"] = TAHUN
fact_pengeluaran = add_kode(pk, sumber="pengeluaran", verbose=False)
fact_pengeluaran = fact_pengeluaran.dropna(
    subset=["pengeluaran_laki_rp", "pengeluaran_perempuan_rp"], how="all")
cek_unik(fact_pengeluaran, ["kode_provinsi", "tahun"], "fact_pengeluaran_kapita")
print("Provinsi terdeteksi di file pengeluaran:", fact_pengeluaran["kode_provinsi"].nunique(), "(harapan: 38 + nasional)")

save_table(fact_miskin, "Sosial", "fact_penduduk_miskin")
save_table(fact_penduduk, "Sosial", "fact_penduduk")
save_table(fact_pengeluaran, "Sosial", "fact_pengeluaran_kapita")
```

```python
# Bencana alam per provinsi (bulanan, detail dampak)
bp = pd.read_excel(find("bencana_dampak"))

# Bersihkan nama kolom agar seragam
bp.columns = [re.sub(r"\W+", "_", str(c).strip().lower()).strip("_") for c in bp.columns]

# Ubah kolom provinsi ke kode_provinsi
bp = bp.rename(columns={"provinsi": "nama_wilayah"})
fact_bencana_prov = add_kode(bp, col="nama_wilayah", sumber="bencana_provinsi")
fact_bencana_prov["tahun"] = TAHUN

# Pastikan tipe data numerik untuk kolom jumlah/dampak
numeric_cols = [c for c in fact_bencana_prov.columns if c not in ["nama_wilayah", "bulan", "jenis_bencana", "kode_provinsi"]]
for c in numeric_cols:
    fact_bencana_prov[c] = to_num(fact_bencana_prov[c]).fillna(0).astype(int)

cek_unik(fact_bencana_prov, ["kode_provinsi", "bulan", "jenis_bencana", "tahun"], "fact_bencana_provinsi")

# Membuat rekapitulasi tingkat nasional secara otomatis dengan aggregasi
rk = fact_bencana_prov.groupby(["jenis_bencana", "tahun"]).sum(numeric_only=True).reset_index()
rk = rk.drop(columns=["kode_provinsi"], errors="ignore")

# Inflasi tahunan (Y-on-Y) bulanan per provinsi
iy = pd.read_excel(find("Inflasi Tahunan"))
fact_inflasi_yoy = add_kode(melt_bulan(iy, "inflasi_yoy_pct"), sumber="inflasi_yoy")
cek_unik(fact_inflasi_yoy, ["kode_provinsi", "bulan_id"], "fact_inflasi_yoy")

save_table(fact_bencana_prov, "Bencana", "fact_bencana_provinsi")
save_table(rk, "Bencana", "fact_bencana_nasional")
save_table(fact_inflasi_yoy, "Inflasi", "fact_inflasi_yoy")
```

```python
def slug(s):
    return re.sub(r"\W+", "_", s.lower()).strip("_")

kom = con.execute("SELECT komoditas_id, nama_komoditas FROM dim_komoditas ORDER BY 1").fetchall()
harga_cols = ",\n        ".join(
    f"MAX(CASE WHEN komoditas_id = {i} THEN harga_rp END) AS harga_{slug(n)}" for i, n in kom)

con.execute(f"""
CREATE OR REPLACE VIEW v_panel_provinsi_bulan AS
WITH harga AS (
    SELECT kode_provinsi, bulan_id,
        {harga_cols}
    FROM fact_harga
    GROUP BY 1, 2
),
bencana AS (
    SELECT kode_provinsi, tahun, SUM(jumlah_kejadian) AS total_bencana
    FROM fact_bencana_provinsi
    GROUP BY 1, 2
)
SELECT
    p.kode_provinsi, p.nama_provinsi, p.wilayah,
    w.bulan_id, w.tahun, w.bulan,
    h.* EXCLUDE (kode_provinsi, bulan_id),
    i.inflasi_yoy_pct,
    t.tpt_pct,
    k.tpak_pct,
    m.persen_miskin, m.jumlah_miskin_ribu, m.garis_kemiskinan_rp,
    u.ump_rp,
    d.jumlah_penduduk_ribu, d.kepadatan_km2,
    b.total_bencana
FROM dim_provinsi p
CROSS JOIN dim_waktu w
LEFT JOIN harga h                ON h.kode_provinsi = p.kode_provinsi AND h.bulan_id = w.bulan_id
LEFT JOIN fact_inflasi_yoy i     ON i.kode_provinsi = p.kode_provinsi AND i.bulan_id = w.bulan_id
LEFT JOIN fact_tpt t             ON t.kode_provinsi = p.kode_provinsi AND t.bulan_id = w.bulan_id
LEFT JOIN fact_tpak k            ON k.kode_provinsi = p.kode_provinsi AND k.bulan_id = w.bulan_id
LEFT JOIN fact_penduduk_miskin m ON m.kode_provinsi = p.kode_provinsi AND m.bulan_id = w.bulan_id
LEFT JOIN fact_ump u             ON u.kode_provinsi = p.kode_provinsi AND u.tahun = w.tahun
LEFT JOIN fact_penduduk d        ON d.kode_provinsi = p.kode_provinsi AND d.tahun = w.tahun
LEFT JOIN bencana b              ON b.kode_provinsi = p.kode_provinsi AND b.tahun = w.tahun
""")

print(con.execute("SHOW TABLES").df())
print(con.execute("""
    SELECT COUNT(*) AS baris, COUNT(DISTINCT kode_provinsi) AS provinsi,
           COUNT(DISTINCT bulan_id) AS bulan
    FROM v_panel_provinsi_bulan""").df())

con.execute("SELECT * FROM v_panel_provinsi_bulan WHERE kode_provinsi = 0 ORDER BY bulan_id").df().head(12)
```
